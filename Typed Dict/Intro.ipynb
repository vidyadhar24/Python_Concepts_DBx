```python
# Standard dict: Accepts any key-value combination
standard_dict = {"username": "admin", 99: "random_value"}


## But, we can specify the type of the data a dictionary object is allowed to access using Typed Dict moudle
## But, this is informational again, passing a different data type still works


from typing import TypedDict

class UserData(TypedDict):
    username : str
    id : int
    is_active : bool




# TypedDict: IDEs and linters enforce this specific structure
valid_user: UserData = {"username": "admin", "id": 101, "is_active": True}
print(valid_user)
```

```python
invalid_user: UserData = {"username": "guest", "id": "one", "role": "viewer"}
print(invalid_user)
```

**Static Type Checking**: this is purely informational.. IDEs and static analysis to such as mypy only catch the errors, but execution doesn't stop

**Runtime Reality:** At runtime, a TypedDict is identical to a standard dict. There is zero performance overhead and absolutely no runtime type enforcement (unlike Pydantic).

**Lightweight Schema**: It acts as a strict blueprint for dictionaries, which is highly useful when passing around structured data like JSON payloads in API or data pipelines without building full class instances.

**Key Constraints**: A standard dict allows any hashable type (like integers or tuples) as a key, but a TypedDict explicitly restricts keys to the predefined strings in your schema.
