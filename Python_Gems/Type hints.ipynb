we know that we need not mention data types before creating a variable in Python..  
We can hint the type just for the sake of information.. i.e., when we hint a type and pass a different one, it still works in Python, but the IDEs and any other external libraries, looking at this information, just flag them as error.

```python
# examples
id: int = 1
name: str = 'Ram'
```

```python
print(id, name)
```

```python
# now, observe this..
cls: int = 'Seventh'
print(cls)
```

**They are purely information**:  They just act as data the act as documentation for IDE, other developers and static to understand the interest of the development.  

**No runtime enforcement**: Python interpreter completely ignores them during the execution, even if we assign different data to declared it would still work. Doesn't throw any error.

**One of the key items here is**:  though Python ignores them for the execution .. it stores these hints in a special dictionary called '__annotations__' which is used by advanced libraries as fast API / Pydantic to perform automated data validation.

```python
print(__annotations__)
```
