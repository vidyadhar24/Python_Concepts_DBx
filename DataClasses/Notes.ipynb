## What are dataclasses ?
- Data classes are advanced dictionaries.
- Dictionary simply holds a key value and finding some value was digging through the complete object.
- But when we create the similar data dictionary using class, we can enforce the schema we can find the objects easily using the methods and on top of that, we can add additional functions to achieve more functionality.
- though we mention the data types, they are just `type hints at run time`.. meaning IDEs and static analysis tools like mypy uses them to format the code
- but, in real time, even if we pass a different data type, it still works and code runs just fine
- if strict data type validation is required, `pydantic` helps here. 
- pydantic `is build on top of dataclasses` and enforce strict data type validations.. resulting in failing the code when a different data type is passed
- one of the key features of using dataclasses is : there three methods are automatically created for any class created with dataclass object
    - __init__
    - __repr__
    - __eq__
- There creates a kind of blueprint for any data class, and that is when we use data class -  when a blueprint is needed to create data objects
- We can simply use the fields needed, apply equality operators among them and even print them.
- using `__repr__`, we normally pass the code used to REPRoduce the object. showing the arguments passed for the devs to debug the issue
- for a class repr must be given, as when used with print(class) or str(class), it falls back to repr(class) when it doesn't have `__str__` method

```python
order_details = {
    'id':1,
    'date':'2025-12-10',
    'amount':200
}
```

```python
from dataclasses import dataclass
from datetime import date

@dataclass
class OrderDetails:    # just use a class name
    id: int            # No __init__ needed, it is one of the points of using dataclasses, gets auto created along with __repr__ & __eq__
    date_: date        # data types are mandatory to be given
    amount: float
    n_items: int

    def get_order_total(self) -> float:
        total = self.n_items * self.amount
        return total
```

```python
o1 = OrderDetails('1', '2026-10-05', 200, 5)
o2 = OrderDetails('2', '2026-11-05', 400, 5)
```

```python
print(o1)
```

```python
print(o1==o2)
```
