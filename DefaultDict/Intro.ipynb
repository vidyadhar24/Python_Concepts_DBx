It just solves the problem of adding key if it doesn't exist yet!

 ![image_1776931635983.png](./image_1776931635983.png "image_1776931635983.png")

```python
boundaries = ['four', 'four', 'six', 'six', 'four']

# to count them by group.. we have to first add the key to normal dict

boundaries_dict = {}

for b in boundaries:
    if b not in boundaries_dict:
        boundaries_dict[b] = 1
    else:
        boundaries_dict[b] += 1
print(boundaries_dict)
```

```python
boundaries = ['four', 'four', 'six', 'six', 'four']

from collections import defaultdict

# we use defaultdict with a predefined datatype.. So it says that if a key doesn't exist and it.. use a default value from that data type as value.. so..for int. it would be 0, for list -> [] )empty list etc.,
# here, we are incrementing the values, other wise first key would be added with 0 as value (ref next cell)

boundaries_dict = defaultdict(int)

for b in boundaries:
    boundaries_dict[b] += 1
print(boundaries_dict)
```

```python
from collections import defaultdict
boundaries = ['four', 'six', 'four']
boundaries_dict = defaultdict(int)

for b in boundaries:
    boundaries_dict[b] += 1
print(boundaries_dict)
```
